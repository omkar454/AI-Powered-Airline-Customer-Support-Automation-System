### Document Cleaning

```python
from pypdf import PdfReader

pdf_path = "knowledge-documents/03_passenger_rights_delays.pdf"
```

```python
reader = PdfReader(pdf_path)
```

```python
clean_text = ""
for i in reader.pages:
    for c in i.extract_text():
        if ord(c) in range(61440, 61696):
            clean_text += chr(ord(c) - 61440)
        else:
            clean_text += c

print(clean_text)
```

```python
print(clean_text)
```

```python
import re
# 1. Clean up excessive empty lines
formatted_body = re.sub(r'\n{3,}', '\n\n', clean_text).strip()
```

```python
print(formatted_body)
```

```python
# 2. Add standard RAG header & footer
rag_header = """DOCUMENT: DGCA PASSENGER CHARTER OF RIGHTS
AUTHORITY: Ministry of Civil Aviation & DGCA, Government of India
CATEGORY: Passenger Rights, Flight Delays, Cancellations, Denied Boarding, and Refunds
SOURCE: Official DGCA Passenger Charter (CAR Section 3, Series M)
==================================================
"""
rag_footer = """
==================================================
END OF DOCUMENT
"""
```

```python
final_document = rag_header + formatted_body + rag_footer
print(final_document)
```

### chunking & vectorization

```python
from langchain_community.document_loaders import  TextLoader, DirectoryLoader

# Load all 5 .txt files

loader = DirectoryLoader(
    path="./knowledge-documents",
    glob = "*.txt",
    loader_cls = TextLoader,
    loader_kwargs = {"encoding" : "utf-8"}
    )

docss = loader.load()

print(len(docss))

for d in docss:
    print(d.metadata["source"])
```

```python
from langchain_text_splitters import  RecursiveCharacterTextSplitter

text_splitters = RecursiveCharacterTextSplitter(
    chunk_size = 600,
    chunk_overlap = 100,
    separators = ["\n\n", "\n", " ", ""]
)

chunks = text_splitters.split_documents(docss)
```

```python
from langchain_community.embeddings import  HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"}
)
```

```python
embedding_model.embed_query("I am Omkar Anand Raut")
```

```python
from langchain_community.vectorstores import  FAISS

vector_db = FAISS.from_documents(chunks, embedding_model)

path = "./vector-db"

vector_db.save_local(path)
```

```python
for i in vector_db.similarity_search("Can I bring an extra laptop bag in the cabin?", k=3):
    print(i)
```
